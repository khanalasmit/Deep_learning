### **What do we normalize?**

We do the normalization on the input layer and the hidden layers.

Beniftis of the normalization are:
- Imporved training stablity by reducing the likelihood of the extreme values
- faster convergence
- mitigating internal covarient shift
    - the input has a certain distribution and when predicting we have different distribution, thus the performance will degrad
    - internal covaient shift is the change of the input for the input layers.
- Regularization effect :
    - such as batch normalization
