### **MULTI HEAD ATTENTION**

The self attention generates the contextual embedding. Suppose we have a sentence "The man saw the astonomer with a telescopt" . It has two possible interpretation one i.e a man used telescope to see astonomer or another the man saw the astonomer haivng a telescope. Self attention generates a single perspective 

 Multi head attention solves this problem. We could use the multi self attention to caputure the multiple meanings. For the multi attention we could use the multiple set of the weights of the query, value, key . 

 

Then we get the multiple contextual representation of the each words.







In the attention is all you need paper we are using the 8 heads and thus 8 architectures of the self attention. The final result is the mixutre of the perspectives.
