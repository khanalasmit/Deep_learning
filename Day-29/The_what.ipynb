**Forget gate**
- consist of the $c_{t-1}, $h_{t-1}, x_t$ as the input
- the number of the nodes in the $\sigma$ let say be 3 layerd.
- Consider the dimension of X as 4 
- the dimesnion of the c_t and h_t would be same as the number of the nodes in the nerual network
- first the f_t is calculated then the pointwise operation of the c_t-1 and the f_t, this means we are forgetting something from the cell state or removing as we are multiplying with f_t
- 
- the weights are $W_{3x7}$, thus 21 weights total and three biases
- then we obtain $f_t$ vector of 3 dimension
- $f_t = \sigma$ $ (W_f[h_{t-1},X_t]+b_f)$

------------------

**Input gate**
- we find new candidates in the cell state, $\tilde{c_t}$
- we then calculate $i_t$
- we then calculate $c_t$
- consist of the $c_{t-1}, h_{t-1}, x_t$ as the input
- the input gate $\;i_t\;$ is calculated by a $\sigma$ neural network just like the forget gate
- the candidate cell state $\;\tilde{c}_t\;$ is calculated with a tanh neural network from the same inputs
- if the dimension of $x_t$ is 4 and the hidden/cell dimension is 3, then the weight matrix for $i_t$ is also $W_{3x7}$ with 21 weights and 3 biases
- the input gate decides what new information to add to the cell state
- finally the cell state is updated as
    - $c_t = f_t * c_{t-1} + i_t * \tilde{c}_t$
- the formulas are:
    - $i_t = \sigma(W_i[h_{t-1}, x_t] + b_i)$
    - $\tilde{c}_t = \tanh(W_c[h_{t-1}, x_t] + b_c)$


---------

**The output gate**
- Output gate decides what part of the cell state should be output as the hidden state
- It uses $h_{t-1}$ and $x_t$ as inputs
- the output gate is calculated by a $\sigma$ neural network just like the forget and input gates
- the hidden state $h_t$ is then computed by gating the cell state with $o_t$
- if the dimension of $x_t$ is 4 and the hidden/cell dimension is 3, then the weight matrix for $o_t$ is also $W_{3x7}$ with 21 weights and 3 biases
- the output gate controls how much of the new cell state is exposed at this time step
- the formulas are:
    - $o_t = \sigma(W_o[h_{t-1}, x_t] + b_o)$
    - $h_t = o_t * \tanh(c_t)$
